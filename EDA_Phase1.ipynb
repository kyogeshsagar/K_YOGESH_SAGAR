```python
# Import Required Libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

# Display graphs inside notebook
%matplotlib inline

# Set plot style
sns.set_style("whitegrid")

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")

print("Libraries Imported Successfully")
```

```python
# Load Dataset

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/SuicideChina.csv"

df = pd.read_csv(url)

print("Dataset Loaded Successfully")
```

```python
# Display First 5 Rows

df.head()
```

```python

# Display Last 5 Rows

df.tail()
```

```python
# Display Random Records

df.sample(5)
```

```python
# Shape of Dataset

rows, columns = df.shape

print("Number of Rows :", rows)
print("Number of Columns :", columns)
```

```python
# Column Names

print(df.columns)
```

```python
# Dataset Information


df.info()
```

```python
# Data Types

df.dtypes
```

```python
# Statistical Summary

df.describe()
```

```python
# Categorical Summary

df.describe(include="object")
```

```python
# Mean

df.mean(numeric_only=True)
```

```python
# Median


df.median(numeric_only=True)
```

```python
# Mode

df.mode().iloc[0]
```

```python
# Minimum Values

df.min(numeric_only=True)
```

```python
# Maximum Values

df.max(numeric_only=True)
```

```python
# Standard Deviation

df.std(numeric_only=True)
```

```python
# Variance

df.var(numeric_only=True)
```

```python
# Quantiles

df.quantile([0.25,0.50,0.75], numeric_only=True)
```

```python
# Skewness

df.skew(numeric_only=True)
```

```python
# Kurtosis

df.kurt(numeric_only=True)
```

```python
# Unique Values Count


df.nunique()
```

```python
# Value Counts for All Categorical Columns

categorical_columns = df.select_dtypes(include='object').columns

for col in categorical_columns:
    print("="*50)
    print("Column :", col)
    print(df[col].value_counts())
    print()
```

```python
# Correlation Matrix

df.corr(numeric_only=True)
```

```python
# Memory Usage

df.memory_usage(deep=True)
```
