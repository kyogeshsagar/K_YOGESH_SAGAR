# 3. Handling Missing Data

Missing values can affect the quality of data analysis and machine learning models. In this section, we identify missing values, calculate their percentage, and handle them appropriately.

```python
# Missing Values in Each Column

df.isnull().sum()
```

```python
# Total Missing Values

total_missing = df.isnull().sum().sum()

print("Total Missing Values :", total_missing)
```

```python
# Percentage of Missing Values

missing_percentage = (df.isnull().sum()/len(df))*100

missing_percentage
```

```python
# Missing Value Summary

missing_summary = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum()/len(df))*100
})

missing_summary
```

```python
# Columns with Missing Values

missing_columns = missing_summary[missing_summary["Missing Values"]>0]

missing_columns
```

```python
# Heatmap of Missing Values

plt.figure(figsize=(12,6))

sns.heatmap(df.isnull(),
            cbar=False,
            cmap="viridis")

plt.title("Missing Values Heatmap")

plt.show()
```

```python
# Missing Value Bar Plot

missing_summary["Missing Values"].plot(
    kind="bar",
    figsize=(12,5)
)

plt.title("Missing Values in Each Column")

plt.ylabel("Count")

plt.show()
```

```python
# Dataset Contains Missing Values?

print(df.isnull().values.any())
```

```python
# Missing Values Per Row

df.isnull().sum(axis=1)
```

```python
# Missing Values Per Column

df.isnull().sum(axis=0)
```

```python
# Fill Numerical Missing Values

numerical_columns = df.select_dtypes(include=np.number).columns

df[numerical_columns] = df[numerical_columns].fillna(
    df[numerical_columns].median()
)

print("Numerical Missing Values Filled")
```

```python
# Fill Categorical Missing Values

categorical_columns = df.select_dtypes(include="object").columns

for col in categorical_columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("Categorical Missing Values Filled")
```

```python
# Verify Missing Values

df.isnull().sum()
```

```python
# Total Missing Values After Handling

print(df.isnull().sum().sum())
```

```python
# Dataset Shape

print(df.shape)
```

```python
# Final Missing Value Report

if df.isnull().sum().sum()==0:
    print("No Missing Values Present in Dataset")
else:
    print("Dataset Still Contains Missing Values")
```

# 5. Data Transformation

Data transformation converts data into a suitable format for analysis. In this section, values are replaced where necessary, new features are created, age is grouped into categories using binning techniques, and outliers are identified.




```python
# Create Copy of Dataset

transformed_df = df.copy()

transformed_df.head()
```

```python
# Replace Blank Values

transformed_df.replace("", np.nan, inplace=True)

print("Blank values replaced with NaN")

transformed_df.head()
```

```python
# Create Age Group

transformed_df["Age_Group"] = pd.cut(
    transformed_df["Age"],
    bins=[0,18,35,60,100],
    labels=["Children","Young Adult","Adult","Senior Citizen"]
)

transformed_df.head()
```

```python
# Age Group Counts

transformed_df["Age_Group"].value_counts()
```

```python
# Age Binning using pd.cut()

transformed_df["Age_Bin"] = pd.cut(
    transformed_df["Age"],
    bins=4
)

transformed_df["Age_Bin"].value_counts()
```

```python
# Age Quartiles

transformed_df["Age_Quartile"] = pd.qcut(
    transformed_df["Age"],
    q=4
)

transformed_df["Age_Quartile"].value_counts()
```

```python
# Create Death Status

transformed_df["Death_Status"] = transformed_df["Died"].replace({
    "yes":"Death",
    "no":"Alive",
    "Yes":"Death",
    "No":"Alive"
})

transformed_df.head()
```

```python
# Hospital Status

transformed_df["Hospital_Status"] = transformed_df["Hospitalised"].replace({
    "yes":"Hospitalised",
    "no":"Not Hospitalised",
    "Yes":"Hospitalised",
    "No":"Not Hospitalised"
})

transformed_df.head()
```

```python
# Outlier Detection using IQR

Q1 = transformed_df["Age"].quantile(0.25)

Q3 = transformed_df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

outliers = transformed_df[
    (transformed_df["Age"] < lower) |
    (transformed_df["Age"] > upper)
]

print("Number of Outliers :", len(outliers))
```

```python
# Display Outliers

outliers
```

# 6. Univariate Analysis

Univariate Analysis is the process of analyzing a single variable to understand its distribution, frequency, spread, and overall characteristics. In this section, different visualization techniques are used to analyze individual variables from the SuicideChina dataset.

```python
# Histogram of Age

plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True,
    color="steelblue"
)

plt.title("Distribution of Age")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.grid(True)

plt.show()
```

```python
# Box Plot of Age

plt.figure(figsize=(8,6))

sns.boxplot(
    y=df["Age"],
    color="lightgreen"
)

plt.title("Box Plot of Age")

plt.ylabel("Age")

plt.show()
```

```python
# Count Plot of Sex

plt.figure(figsize=(8,6))

sns.countplot(
    data=df,
    x="Sex",
    palette="Set2"
)

plt.title("Count of Individuals by Sex")

plt.xlabel("Sex")

plt.ylabel("Count")

plt.show()
```

```python
# Count Plot of Occupation

plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x="Occupation",
    order=df["Occupation"].value_counts().index,
    palette="viridis"
)

plt.xticks(rotation=45)

plt.title("Occupation Distribution")

plt.xlabel("Occupation")

plt.ylabel("Count")

plt.show()
```

```python
# Count Plot of Suicide Method

plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x="method",
    order=df["method"].value_counts().index,
    palette="magma"
)

plt.xticks(rotation=45)

plt.title("Distribution of Suicide Methods")

plt.xlabel("Method")

plt.ylabel("Count")

plt.show()
```

```python
# Count Plot of Education

plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x="Education",
    order=df["Education"].value_counts().index,
    palette="coolwarm"
)

plt.xticks(rotation=45)

plt.title("Education Distribution")

plt.xlabel("Education")

plt.ylabel("Count")

plt.show()
```

# 8. Multivariate Analysis

Multivariate Analysis examines the relationship among three or more variables simultaneously. It helps identify complex patterns, correlations, and interactions within the dataset through advanced visualization techniques.

```python
# Correlation Heatmap

plt.figure(figsize=(8,6))

correlation_matrix = df.select_dtypes(include=['int64','float64']).corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Pair Plot

sns.pairplot(
    df[['Age','Year','Month']],
    diag_kind="hist"
)

plt.show()
```

```python
# Scatter Plot with Hue

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="Year",
    y="Age",
    hue="Sex",
    palette="Set1"
)

plt.title("Age vs Year by Sex")

plt.xlabel("Year")

plt.ylabel("Age")

plt.legend(title="Sex")

plt.show()
```

```python
# Box Plot

plt.figure(figsize=(14,6))

sns.boxplot(
    data=df,
    x="Occupation",
    y="Age",
    hue="Sex"
)

plt.xticks(rotation=45)

plt.title("Age Distribution by Occupation and Sex")

plt.xlabel("Occupation")

plt.ylabel("Age")

plt.show()
```

```python
# Grouped Bar Plot

plt.figure(figsize=(12,6))

sns.barplot(
    data=df,
    x="Education",
    y="Age",
    hue="Sex",
    estimator="mean"
)

plt.xticks(rotation=45)

plt.title("Average Age by Education and Sex")

plt.xlabel("Education")

plt.ylabel("Average Age")

plt.show()
```
